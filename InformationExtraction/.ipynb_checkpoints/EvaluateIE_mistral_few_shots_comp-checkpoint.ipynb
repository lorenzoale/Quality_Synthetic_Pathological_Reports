### helpers fc

```python
NUMERIC_FIELDS = {
    "numero_gang_prel",
    "numero_gang_att",
    "taille",
    "ki67",
}

def is_missing(x):
    if x is None:
        return True

    # Avoid pd.isna errors on lists/dicts
    try:
        if pd.isna(x):
            return True
    except Exception:
        pass

    if isinstance(x, str):
        x_clean = x.strip().lower()
        if x_clean in ["", "unknown", "nan", "none", "null"]:
            return True

    return False

def normalize_text(x):
    if is_missing(x):
        return "unknown"

    x = str(x).strip().lower()
    x = " ".join(x.split())

    # Remove accents if needed
    replacements = {
        "é": "e",
        "è": "e",
        "ê": "e",
        "ë": "e",
        "à": "a",
        "â": "a",
        "î": "i",
        "ï": "i",
        "ô": "o",
        "ö": "o",
        "ù": "u",
        "û": "u",
        "ü": "u",
        "ç": "c",
    }

    for a, b in replacements.items():
        x = x.replace(a, b)

    return x
```

```python
def parse_numeric_bin(x):
    """
    Converts gold labels like:
    '1-5', '6-10', '11-20', '21-50', '41-100'
    '+10', '50+'
    '0'
    into numeric intervals.
    """
    if is_missing(x):
        return None

    text = str(x).strip().lower().replace(",", ".")

    # 50+ means >= 50
    match = re.match(r"^(\d+(?:\.\d+)?)\s*\+$", text)
    if match:
        low = float(match.group(1))
        return low, float("inf")

    # +10 means >= 10
    match = re.match(r"^\+\s*(\d+(?:\.\d+)?)$", text)
    if match:
        low = float(match.group(1))
        return low, float("inf")

    # 1-5, 6-10, 11-20, 21-50, 41-100
    match = re.match(r"^(\d+(?:\.\d+)?)\s*[-–—]\s*(\d+(?:\.\d+)?)$", text)
    if match:
        low = float(match.group(1))
        high = float(match.group(2))
        return min(low, high), max(low, high)

    # Exact numeric bin: 0
    match = re.match(r"^\d+(?:\.\d+)?$", text)
    if match:
        value = float(text)
        return value, value

    return None



def extract_pred_number(x):
    if is_missing(x):
        return None

    if isinstance(x, (int, float, np.integer, np.floating)):
        if pd.isna(x):
            return None
        return float(x)

    match = re.search(r"-?\d+(?:[\.,]\d+)?", str(x))
    if match:
        return float(match.group(0).replace(",", "."))

    return None


def numeric_bin_correct(attr, gold_option, pred_value):
    pred_num = extract_pred_number(pred_value)

    if pred_num is None:
        return False

    interval = parse_numeric_bin(gold_option)

    if interval is None:
        return False

    low, high = interval

    # Annotation hack:
    # Ki67 predicted 40 should count as correct for gold 41-100
    # if attr == "ki67" and pred_num == 40 and low == 41 and high == 100:
    #     return True
    if attr == "ki67":
        return (low - 1) <= pred_num <= high
        
    return low <= pred_num <= high

    
# def extract_number(x):
#     if is_missing(x):
#         return None

#     if isinstance(x, (int, float, np.integer, np.floating)):
#         if pd.isna(x):
#             return None
#         return float(x)

#     match = re.search(r"-?\d+(?:[\.,]\d+)?", str(x))
#     if match:
#         return float(match.group(0).replace(",", "."))

#     return None


# def extract_range(x):
#     if is_missing(x):
#         return None

#     text = str(x).lower().replace(",", ".")

#     # Match ranges: 41-100, 41 – 100, 41 à 100, 41 to 100
#     match = re.search(
#         r"(-?\d+(?:\.\d+)?)\s*(?:-|–|—|à|a|to)\s*(-?\d+(?:\.\d+)?)",
#         text
#     )

#     if match:
#         low = float(match.group(1))
#         high = float(match.group(2))
#         return min(low, high), max(low, high)

#     return None


# def numeric_correct(attr, gold_value, pred_value):
#     pred_num = extract_number(pred_value)

#     if pred_num is None:
#         return False

#     gold_range = extract_range(gold_value)

#     # Gold is a range
#     if gold_range is not None:
#         low, high = gold_range

#         # Specific annotation hack:
#         # Ki67 pred 40 should count as correct if gold is 41-100
#         if attr == "ki67" and pred_num == 40 and low == 41 and high == 100:
#             return True

#         return low <= pred_num <= high

#     # Gold is a single number
#     gold_num = extract_number(gold_value)

#     if gold_num is None:
#         return False

#     return pred_num == gold_num
```

```python
def normalize_diagnostic(x):
    x = normalize_text(x)

    diagnostic_synonyms = {
        "carcinome canalaire infiltrant": "adenocarcinome canalaire infiltrant",
        "adenocarcinome canalaire infiltrant": "adenocarcinome canalaire infiltrant",
        "carcinome infiltrant de type non specifique": "adenocarcinome canalaire infiltrant",
        "carcinome mammaire infiltrant de type non specifique": "adenocarcinome canalaire infiltrant",
    }

    return diagnostic_synonyms.get(x, x)
```

### Attribute-specific correctness

```python
def checkCorrect(attr, gold_value, pred_value):
    # Both missing = correct
    if is_missing(gold_value) and is_missing(pred_value):
        return True

    # One missing but not the other = incorrect
    if is_missing(gold_value) != is_missing(pred_value):
        return False

    if attr in NUMERIC_FIELDS:
        gold_options = [
            option.strip()
            for option in str(gold_value).split(";")
            if option.strip() != ""
        ]

        return any(
            numeric_bin_correct(attr, option, pred_value)
            for option in gold_options
        )

    # Text fields
    if attr == "diagnostic":
        norm_pred = normalize_diagnostic(pred_value)
        gold_options = [
            normalize_diagnostic(option)
            for option in str(gold_value).split(";")
            if option.strip() != ""
        ]
    else:
        norm_pred = normalize_text(pred_value)
        gold_options = [
            normalize_text(option)
            for option in str(gold_value).split(";")
            if option.strip() != ""
        ]

    return norm_pred in gold_options
```

### Convert results_base to long evaluation DataFrame

```python
eval_rows = []
FIELD_ORDER = [
    "echantillon",
    "numero_gang_prel",
    "numero_gang_att",
    "taille",
    "diagnostic",
    "grade",
    "emboles",
    "ki67",
    "re",
    "rp",
    "her2",
    "marges_saines",
    "rupture_capsulaire",
]


for row_idx, item in results_base_df.iterrows():
    # row_id = item["row_id"] if "row_id" in results_base_df.columns else row_idx
    row_id = row_idx

    gold = item["gold_variables"] if isinstance(item["gold_variables"], dict) else {}
    pred = item["prediction"] if isinstance(item["prediction"], dict) else {}

    for attr in FIELD_ORDER:
        gold_value = gold.get(attr)
        pred_value = pred.get(attr)

        eval_rows.append({
            "row_id": row_id,
            "attribute": attr,
            "gold_value": gold_value,
            "pred_value": pred_value,
            "value_correct": checkCorrect(attr, gold_value, pred_value),
        })

df_eval = pd.DataFrame(eval_rows)
df_eval.head(13)
```

Patch the known annotation bug in first row

```python
first_row_id = df_eval["row_id"].iloc[0]

df_eval.loc[
    (df_eval["row_id"] == first_row_id) & (df_eval["attribute"] == "marges_saines"),
    "gold_value"
] = "oui"

df_eval.loc[
    (df_eval["row_id"] == first_row_id) & (df_eval["attribute"] == "rupture_capsulaire"),
    "gold_value"
] = "non"

# Recompute correctness after patch
df_eval["value_correct"] = df_eval.apply(
    lambda row: checkCorrect(row["attribute"], row["gold_value"], row["pred_value"]),
    axis=1
)
```

### Compute attribute-level metrics

```python
rows = []

for attr, group in df_eval.groupby("attribute"):
    gold_missing = group["gold_value"].apply(is_missing)
    pred_missing = group["pred_value"].apply(is_missing)

    has_gold = ~gold_missing
    has_pred = ~pred_missing
    correct = group["value_correct"]

    TP = has_pred & correct
    FP = has_pred & ~correct
    FN = has_gold & ~has_pred

    tp_sum = TP.sum()
    fp_sum = FP.sum()
    fn_sum = FN.sum()

    precision = tp_sum / (tp_sum + fp_sum) if (tp_sum + fp_sum) > 0 else float("nan")
    recall = tp_sum / (tp_sum + fn_sum) if (tp_sum + fn_sum) > 0 else float("nan")
    f1 = (
        2 * precision * recall / (precision + recall)
        if precision + recall > 0
        else float("nan")
    )

    rows.append({
        "attribute": attr,
        "tp": tp_sum,
        "fp": fp_sum,
        "fn": fn_sum,
        "precision": precision,
        "recall": recall,
        "f1": f1,
    })

df_attr_metrics = pd.DataFrame(rows)
df_attr_metrics
```

```python
TP_total = df_attr_metrics["tp"].sum()
FP_total = df_attr_metrics["fp"].sum()
FN_total = df_attr_metrics["fn"].sum()

micro_precision = TP_total / (TP_total + FP_total) if (TP_total + FP_total) > 0 else float("nan")
micro_recall = TP_total / (TP_total + FN_total) if (TP_total + FN_total) > 0 else float("nan")
micro_f1 = (
    2 * micro_precision * micro_recall / (micro_precision + micro_recall)
    if micro_precision + micro_recall > 0
    else float("nan")
)

overall_metrics = {
    "tp": TP_total,
    "fp": FP_total,
    "fn": FN_total,
    "precision": micro_precision,
    "recall": micro_recall,
    "f1": micro_f1,
}

overall_metrics
```

### Macro and Micro acc and per att all toghether

```python
import numpy as np
import pandas as pd

def compute_metrics(df_eval, exclude_attrs=None):
    if exclude_attrs is None:
        exclude_attrs = []

    df_use = df_eval[~df_eval["attribute"].isin(exclude_attrs)].copy()

    rows = []

    for attr, group in df_use.groupby("attribute"):
        gold_missing = group["gold_value"].apply(is_missing)
        pred_missing = group["pred_value"].apply(is_missing)

        has_gold = ~gold_missing
        has_pred = ~pred_missing
        correct = group["value_correct"]

        TP = has_pred & correct
        FP = has_pred & ~correct
        FN = has_gold & ~has_pred

        tp_sum = TP.sum()
        fp_sum = FP.sum()
        fn_sum = FN.sum()

        precision = tp_sum / (tp_sum + fp_sum) if (tp_sum + fp_sum) > 0 else np.nan
        recall = tp_sum / (tp_sum + fn_sum) if (tp_sum + fn_sum) > 0 else np.nan
        f1 = (
            2 * precision * recall / (precision + recall)
            if pd.notna(precision) and pd.notna(recall) and (precision + recall) > 0
            else np.nan
        )

        # Accuracy for this attribute:
        # number correct / number evaluated
        accuracy = correct.mean()

        rows.append({
            "attribute": attr,
            "n": len(group),
            "correct": correct.sum(),
            "accuracy": accuracy,
            "tp": tp_sum,
            "fp": fp_sum,
            "fn": fn_sum,
            "precision": precision,
            "recall": recall,
            "f1": f1,
        })

    df_attr_metrics = pd.DataFrame(rows)

    TP_total = df_attr_metrics["tp"].sum()
    FP_total = df_attr_metrics["fp"].sum()
    FN_total = df_attr_metrics["fn"].sum()

    micro_precision = TP_total / (TP_total + FP_total) if (TP_total + FP_total) > 0 else np.nan
    micro_recall = TP_total / (TP_total + FN_total) if (TP_total + FN_total) > 0 else np.nan
    micro_f1 = (
        2 * micro_precision * micro_recall / (micro_precision + micro_recall)
        if pd.notna(micro_precision) and pd.notna(micro_recall) and (micro_precision + micro_recall) > 0
        else np.nan
    )

    overall_metrics = {
        "excluded_attributes": exclude_attrs,
        "n_attributes": df_attr_metrics["attribute"].nunique(),
        "n_rows": len(df_use),

        # Global accuracy over all attribute-row pairs
        "micro_accuracy": df_use["value_correct"].mean(),

        # Average of per-attribute accuracies
        "macro_accuracy": df_attr_metrics["accuracy"].mean(),

        # Optional weighted average of per-attribute accuracies
        "weighted_accuracy": (
            (df_attr_metrics["accuracy"] * df_attr_metrics["n"]).sum()
            / df_attr_metrics["n"].sum()
            if df_attr_metrics["n"].sum() > 0 else np.nan
        ),

        "tp": TP_total,
        "fp": FP_total,
        "fn": FN_total,
        "precision": micro_precision,
        "recall": micro_recall,
        "f1": micro_f1,
    }

    return df_attr_metrics, overall_metrics
```

```python
df_attr_metrics, overall_metrics = compute_metrics(df_eval)

df_attr_metrics
```

```python
def metrics_table_to_latex(df_attr_metrics, overall_metrics, path="attribute_metrics.tex"):
    table = df_attr_metrics.copy()

    # Optional formatting
    metric_cols = ["accuracy", "precision", "recall", "f1"]
    for col in metric_cols:
        if col in table.columns:
            table[col] = table[col].round(2)

    # Add Total row at the end
    total_row = {
        "attribute": "Total",
        # "n": overall_metrics.get("n_rows", table["n"].sum() if "n" in table.columns else None),
        # "correct": table["correct"].sum() if "correct" in table.columns else None,
        "accuracy": round(overall_metrics.get("micro_accuracy", float("nan")), 2),
        # "tp": overall_metrics.get("tp"),
        # "fp": overall_metrics.get("fp"),
        # "fn": overall_metrics.get("fn"),
        "precision": round(overall_metrics.get("precision", float("nan")), 2),
        "recall": round(overall_metrics.get("recall", float("nan")), 2),
        "f1": round(overall_metrics.get("f1", float("nan")), 2),
    }

    table = pd.concat([table, pd.DataFrame([total_row])], ignore_index=True)

    # Keep only columns that exist
    preferred_cols = [
        "attribute",
        # "n",
        # "correct",
        "accuracy",
        # "tp",
        # "fp",
        # "fn",
        "precision",
        "recall",
        "f1",
    ]
    table = table[[c for c in preferred_cols if c in table.columns]]

    latex = table.to_latex(
        index=False,
        escape=True,
        float_format="%.2f",
        caption="Attribute-level extraction metrics.",
        label="tab:attribute_metrics",
    )

    with open(path, "w", encoding="utf-8") as f:
        f.write(latex)

    return latex, table

latex, table_with_total = metrics_table_to_latex(
    df_attr_metrics,
    overall_metrics,
    path="attribute_metrics.tex"
)

print(latex)
```
