# Mixtral 8x7B Instruct / MIstral 7b — base few-shot pathology extraction

This notebook runs **base** `mistralai/Mixtral-8x7B-Instruct-v0.1` for pathology-report JSON extraction.

It mirrors the MedGemma inference notebook but uses **few-shot prompting** instead of a LoRA adapter.

Key points:
- No adapter is loaded.
- Few-shot examples are inserted as alternating `user` / `assistant` messages.
- Batch inference uses left padding.
- Prompt tokenization uses `truncation=False` so reports are not silently cut.
- JSON parsing uses explicit brace-matching extraction.

## 3. Field schema and task instruction

```python
FIELD_ORDER = [
    "echantillon",
    "numero_gang_prel",
    "numero_gang_att",
    "taille",
    "diagnostic",
    "grade",
    "emboles",
    "ki67",
    "re",
    "rp",
    "her2",
    "marges_saines",
    "rupture_capsulaire",
]

NUMERIC_FIELDS = {"numero_gang_prel", "numero_gang_att", "taille", "ki67"}

SYSTEM_PROMPT = """
Tu es un système d'extraction d'information à partir de comptes rendus anatomopathologiques de cancer du sein.

Retourne uniquement un objet JSON valide.
N'invente aucune information.

Si une information est absente, ambiguë ou non déductible:
- utilise "unknown" pour les variables catégorielles
- utilise null pour les variables numériques.

Variables à extraire:
- echantillon: mammectomie, tumorectomie, curage ganglionnaire, biopsie, exérèse chirurgicale, unknown
- numero_gang_prel: nombre de ganglions prélevés/examinés, integer ou null
- numero_gang_att: nombre de ganglions atteints/envahis, integer ou null
- taille: taille tumorale maximale en mm, integer ou null
- diagnostic: histologie principale, string
- grade: SBR1, SBR2, SBR3, unknown
- emboles: oui, non, unknown
- ki67: pourcentage Ki67, integer ou null
- re: positifs, négatifs, unknown
- rp: positifs, négatifs, unknown
- her2: négatif, douteux, positif, unknown
- marges_saines: oui, non, unknown
- rupture_capsulaire: oui, non, unknown

Règles:
- Convertir les cm en mm pour taille.
- HER2 0 ou 1+ = négatif.
- HER2 2+ = douteux.
- HER2 3+ ou amplification FISH/CISH/SISH = positif.
- RE/RP positifs si expression positive, même faible.
- Marges R0/non envahies/saines = oui.
- Marges atteintes/R1/envahies = non.
""".strip()
```

## 4. Few-shot examples

Use a small number of examples, usually 2-5. Keep them short and representative.

Important: do **not** use the same rows for few-shot examples and evaluation if you want a clean benchmark.

```python
TRAIN_FILE = "training_data/train_normalized.jsonl"
N_SHOTS=3
SEED=42
def load_jsonl_examples(path: str, max_examples: Optional[int] = None) -> List[Dict[str, Any]]:
    examples = []
    with open(path, "r", encoding="utf-8") as f:
        for i, line in enumerate(f):
            if max_examples is not None and i >= max_examples:
                break
            ex = json.loads(line)
            if "report" not in ex or "label" not in ex:
                raise ValueError(f"Each JSONL row must have report and label keys. Bad row index: {i}")
            examples.append(ex)
    return examples



train_examples = load_jsonl_examples(TRAIN_FILE)
print("Loaded train examples:", len(train_examples))
print("First example keys:", train_examples[0].keys())
```

```python
def examples_to_train_df(train_examples: List[Dict[str, Any]]) -> pd.DataFrame:
    """Create a dataframe with one row per training example.

    The dataframe index is the same as the train_examples list position, so
    train_df.index values can be used directly as few-shot indices.
    """
    rows = []
    for i, ex in enumerate(train_examples):
        label = parse_label(ex["label"])
        row = {"row_id": i, "report": ex["report"]}
        for field in FIELD_ORDER:
            value = label.get(field)
            if value == "unknown" and field in NUMERIC_FIELDS:
                value = None
            row[field] = value
        rows.append(row)

    train_df = pd.DataFrame(rows)
    train_df.index = train_df["row_id"].astype(int)

    # Make numeric fields numeric so comparisons such as > 0 work reliably.
    for field in NUMERIC_FIELDS:
        train_df[field] = pd.to_numeric(train_df[field], errors="coerce")

    return train_df



from typing import Dict, List
import pandas as pd

def parse_label(label: Any) -> Dict[str, Any]:
    if isinstance(label, str):
        label = json.loads(label)
    if not isinstance(label, dict):
        raise ValueError(f"label must be a dict or JSON string, got {type(label)}")
    return label

def dumps_label(label: Dict[str, Any]) -> str:
    ordered = {}
    for key in FIELD_ORDER:
        value = label.get(key)
        if key in NUMERIC_FIELDS:
            if value == "unknown" or (isinstance(value, float) and math.isnan(value)):
                value = None
            elif value is not None:
                try:
                    value = int(value)
                except (TypeError, ValueError):
                    pass
        else:
            if value is None or (isinstance(value, float) and math.isnan(value)):
                value = "unknown"
        ordered[key] = value
    return json.dumps(ordered, ensure_ascii=False, separators=(",", ":"))
    
def find_candidates(train_df: pd.DataFrame) -> Dict[str, List[int]]:
    candidates = {}

    def s(col):
        return train_df[col].astype(str).str.strip().str.lower()

    # 1. Biopsie with in situ:
    #    diagnostic 'carcinome intracanalaire non infiltrant'
    mask_biopsie_absent = (
        s("echantillon").eq("biopsie")
        & (
            train_df["diagnostic"].isna()
            | s("diagnostic").isin(["", "unknown", "nan", "none", "null", 'carcinome intracanalaire non infiltrant'])
        )
        & train_df["numero_gang_prel"].isna()
        & train_df["numero_gang_att"].isna()
        & train_df["taille"].isna()
        & train_df["ki67"].isna()
        & s("grade").eq("unknown")
    )

    candidates["biopsie_absent_information"] = (
        train_df[mask_biopsie_absent].index.astype(int).tolist()[:10]
    )

    # 2. Tumorectomie with ganglions:
    #    known sampled/examined nodes and known positive/affected nodes
    mask_tumorectomie_ganglions = (
        s("echantillon").eq("tumorectomie")
        & train_df["numero_gang_prel"].notna()
        & train_df["numero_gang_att"].notna()
        & (train_df["numero_gang_prel"] > 0)
    )

    candidates["tumorectomie_with_ganglions"] = (
        train_df[mask_tumorectomie_ganglions].index.astype(int).tolist()[:10]
    )

    # 3. Mammectomie:
    #    preferably with missing ganglions / missing Ki67 / unknown grade,
    #    to keep the example useful for absent values too
    mask_mammectomie = (
        s("echantillon").eq("mammectomie")
        & train_df["numero_gang_prel"].isna()
        & train_df["numero_gang_att"].isna()
        & train_df["ki67"].isna()
        & s("grade").eq("unknown")
    )

    candidates["mammectomie_missing_values"] = (
        train_df[mask_mammectomie].index.astype(int).tolist()[:10]
    )

    return candidates

def choose_representative_shots(
    train_examples: List[Dict[str, Any]],
    n_shots: int = 3,
    seed: int = 42,
    category_order: Optional[List[str]] = None,
) -> List[int]:
    """Choose fixed clinically representative shots.

    It tries to pick one example from each category in category_order.
    If a category has no candidate, it skips it. If fewer than n_shots are
    found, it fills the rest with random unused examples.
    """
    rng = random.Random(seed)
    train_df = examples_to_train_df(train_examples)
    candidates = find_candidates(train_df)

    if category_order is None:
        category_order = [
            "mammectomie_missing_values",
            "biopsie_absent_information",
            "tumorectomie_with_ganglions",
            "her2_douteux",
            "marges_non_saines",
        ]

    selected = []
    selected_from = []

    for category in category_order:
        available = [idx for idx in candidates.get(category, []) if idx not in selected]
        if not available:
            continue
        selected.append(rng.choice(available))
        selected_from.append(category)
        if len(selected) >= n_shots:
            break

    if len(selected) < n_shots:
        remaining = [i for i in range(len(train_examples)) if i not in selected]
        fill = rng.sample(remaining, k=min(n_shots - len(selected), len(remaining)))
        selected.extend(fill)
        selected_from.extend(["random_fallback"] * len(fill))

    print("Candidate counts:")
    for name, idxs in candidates.items():
        print(f"- {name}: {len(idxs)} shown / first candidates: {idxs[:5]}")

    print("\nSelected shots:")
    for idx, source in zip(selected, selected_from):
        print(f"- {idx} from {source}: {dumps_label(parse_label(train_examples[idx]['label']))}")

    return selected


# Option A: fixed manual indices
# shot_indices = [0, 1, 2]

# Option B: clinically representative fixed shots
shot_indices = choose_representative_shots(
    train_examples,
    n_shots=N_SHOTS,
    seed=SEED,
)

print("\nshot_indices:", shot_indices)
```

```python
def ordered_label(label: Dict[str, Any]) -> Dict[str, Any]:
    """Return the label with stable field order and numeric unknown -> None."""
    out = {}
    for field in FIELD_ORDER:
        value = label.get(field, "unknown")
        if field in NUMERIC_FIELDS and value == "unknown":
            value = None
        out[field] = value
    return out


def few_shots_from_df(df_source, row_indices):
    shots = []

    for row_idx in row_indices:
        # Works for either pandas DataFrame or list of dicts
        if hasattr(df_source, "iloc"):
            row = df_source.iloc[row_idx]
            get_value = lambda key: row[key] if key in row else None
        else:
            row = df_source[row_idx]
            get_value = lambda key: row.get(key, None)

        label = {
            field: get_value(field)
            for field in FIELD_ORDER
        }

        clean_label = {}
        for field, value in label.items():
            if field in NUMERIC_FIELDS:
                if value is None or value == "" or str(value).lower() == "unknown":
                    clean_label[field] = None
                else:
                    clean_label[field] = int(value)
            else:
                if value is None or value == "":
                    clean_label[field] = "unknown"
                else:
                    clean_label[field] = str(value)

        shots.append({
            "report": get_value("content"),
            "label": ordered_label(clean_label),
        })

    return shots


FEW_SHOTS = few_shots_from_df(train_examples, shot_indices)
```

## 5. JSON parser

```python
def extract_first_json_object(text: str) -> Optional[Dict[str, Any]]:
    """Best-effort extraction of the first JSON object from model output."""
    start = text.find("{")
    if start == -1:
        return None

    depth = 0
    in_string = False
    escape = False

    for idx in range(start, len(text)):
        char = text[idx]

        if escape:
            escape = False
            continue
        if char == "\\":
            escape = True
            continue
        if char == '"':
            in_string = not in_string
            continue
        if in_string:
            continue
        if char == "{":
            depth += 1
        elif char == "}":
            depth -= 1
            if depth == 0:
                candidate = text[start:idx + 1]
                try:
                    parsed = json.loads(candidate)
                    return parsed if isinstance(parsed, dict) else None
                except json.JSONDecodeError:
                    return None

    return None
```

## 6. Prompt builder for Mixtral few-shot chat

The examples are alternating `user` and `assistant` messages. The final real report is a `user` message, and `add_generation_prompt=True` tells Mixtral to generate the assistant completion.

```python
# def build_report_user_message(report: str) -> str:
#     return f"""Compte rendu:
# \"\"\"
# {report}
# \"\"\"

# Retourne uniquement le JSON."""
def build_report_user_message(report: str) -> str:
    return f"""{SYSTEM_PROMPT}

Compte rendu:
\"\"\"
{report}
\"\"\"

Retourne uniquement le JSON."""


def build_mixtral_fewshot_prompt(
    report: str,
    tokenizer: Any,
    few_shots: List[Dict[str, Any]] = FEW_SHOTS,
) -> str:
    messages = []

    # # For Mixtral Instruct, this is robust: put global instructions inside user content.
    # messages.append({
    #     "role": "user",
    #     "content": SYSTEM_PROMPT + "\n\nVoici des exemples. Réponds ensuite au dernier compte rendu avec uniquement le JSON.",
    # })
    # messages.append({
    #     "role": "assistant",
    #     "content": "Compris. Je retournerai uniquement un objet JSON valide.",
    # })

    for ex in few_shots:
        messages.append({
            "role": "user",
            "content": build_report_user_message(ex["report"]),
        })
        messages.append({
            "role": "assistant",
            "content": json.dumps(
                ordered_label(ex["label"]),
                ensure_ascii=False,
                separators=(",", ":"),
            ),
        })

    messages.append({
        "role": "user",
        "content": build_report_user_message(report),
    })

    return tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True,
    )
```

## 7. Load tokenizer and model

```python
tokenizer = AutoTokenizer.from_pretrained(MODEL_ID_OR_PATH, use_fast=True)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

tokenizer.padding_side = "left"

model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID_OR_PATH,
    torch_dtype=DTYPE,
    device_map=DEVICE_MAP,
)
model.eval()
```

```python
def get_input_device(model: Any) -> torch.device:
    """Choose a safe device for input tensors when using device_map='auto'."""
    try:
        return model.device
    except Exception:
        return next(model.parameters()).device


def count_prompt_tokens(report: str, tokenizer: Any, few_shots: List[Dict[str, Any]] = FEW_SHOTS) -> int:
    prompt = build_mixtral_fewshot_prompt(report, tokenizer, few_shots=few_shots)
    return len(tokenizer(prompt, add_special_tokens=False)["input_ids"])

# Quick check on the first report.
count_prompt_tokens(df["content"].iloc[0], tokenizer)
```

## 8. Single-report inference

```python
@torch.inference_mode()
def generate_mixtral_extraction(
    report: str,
    model: Any,
    tokenizer: Any,
    few_shots: List[Dict[str, Any]] = FEW_SHOTS,
    max_new_tokens: int = MAX_NEW_TOKENS,
    return_raw: bool = True,
    return_prompt: bool = False,
) -> Dict[str, Any]:
    prompt = build_mixtral_fewshot_prompt(report, tokenizer, few_shots=few_shots)

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=False,
    )
    inputs = {k: v.to(get_input_device(model)) for k, v in inputs.items()}

    generation_config = GenerationConfig(
        max_new_tokens=max_new_tokens,
        do_sample=False,
        pad_token_id=tokenizer.pad_token_id,
        eos_token_id=tokenizer.eos_token_id,
    )

    output_ids = model.generate(
        **inputs,
        generation_config=generation_config,
    )

    generated_ids = output_ids[0][inputs["input_ids"].shape[-1]:]
    raw_text = tokenizer.decode(generated_ids, skip_special_tokens=True)
    prediction = extract_first_json_object(raw_text)

    result = {"prediction": prediction}
    if return_raw:
        result["raw_text"] = raw_text
    if return_prompt:
        result["prompt"] = prompt
    return result
```

```python
result_one = generate_mixtral_extraction(
    report=df["content"].iloc[0],
    model=model,
    tokenizer=tokenizer,
    max_new_tokens=MAX_NEW_TOKENS,
    return_raw=True,
    return_prompt=True,
)

result_one["prediction"]
```

## 9. Batch inference

```python
@torch.inference_mode()
def generate_mixtral_extraction_batch(
    reports: List[str],
    model: Any,
    tokenizer: Any,
    few_shots: List[Dict[str, Any]] = FEW_SHOTS,
    batch_size: int = BATCH_SIZE,
    max_new_tokens: int = MAX_NEW_TOKENS,
) -> List[Dict[str, Any]]:
    old_padding_side = tokenizer.padding_side
    tokenizer.padding_side = "left"

    results: List[Dict[str, Any]] = []
    input_device = get_input_device(model)

    try:
        for start in tqdm(range(0, len(reports), batch_size), desc="Running Mixtral batch inference"):
            batch_reports = reports[start:start + batch_size]
            prompts = [
                build_mixtral_fewshot_prompt(report, tokenizer, few_shots=few_shots)
                for report in batch_reports
            ]

            inputs = tokenizer(
                prompts,
                return_tensors="pt",
                padding=True,
                truncation=False,
            )
            inputs = {k: v.to(input_device) for k, v in inputs.items()}

            generation_config = GenerationConfig(
                max_new_tokens=max_new_tokens,
                do_sample=False,
                pad_token_id=tokenizer.pad_token_id,
                eos_token_id=tokenizer.eos_token_id,
            )

            output_ids = model.generate(
                **inputs,
                generation_config=generation_config,
            )

            prompt_width = inputs["input_ids"].shape[-1]
            for i, report in enumerate(batch_reports):
                generated_ids = output_ids[i][prompt_width:]
                raw_text = tokenizer.decode(generated_ids, skip_special_tokens=True)
                pred = extract_first_json_object(raw_text)
                results.append({
                    "raw_text": raw_text,
                    "prediction": pred,
                })

    finally:
        tokenizer.padding_side = old_padding_side

    return results
```

## 10. Run Mixtral few-shot inference on the dataframe

```python
reports = df["content"].astype(str).tolist()
batch_outputs = generate_mixtral_extraction_batch(
    reports=reports,
    model=model,
    tokenizer=tokenizer,
    few_shots=FEW_SHOTS,
    batch_size=BATCH_SIZE,
    max_new_tokens=MAX_NEW_TOKENS,
)

results_mixtral = []
for i, (report, out) in enumerate(zip(reports, batch_outputs)):
    gold_variables = df.iloc[i][gold_cols].to_dict()
    results_mixtral.append({
        "row_id": i,
        "report": report,
        "gold_variables": gold_variables,
        "prediction": out["prediction"],
        "raw_text": out["raw_text"],
    })

results_mixtral_df = pd.DataFrame(results_mixtral)
results_mixtral_df.head()
```
